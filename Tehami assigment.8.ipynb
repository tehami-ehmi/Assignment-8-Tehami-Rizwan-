# Project: Online Shopping Cart

```python
products = {
    "P1": {"name": "Shoes",      "price": 2500, "stock": 5},
    "P2": {"name": "Shirt",      "price": 1500, "stock": 10},
    "P3": {"name": "Bag",        "price": 3000, "stock": 4},
    "P4": {"name": "Watch",      "price": 4000, "stock": 3},
    "P5": {"name": "Headphones", "price": 2000, "stock": 6},
}

cart = {}         

def show_menu():
    print("\n" + "-" * 40)
    print("       ONLINE SHOPPING CART")
    print("-" * 40)
    print("1. View Products")
    print("2. Add to Cart")
    print("3. View Cart")
    print("4. Checkout")
    print("5. Exit")
    print("-" * 40)


def view_products():
    print("\n" + "=" * 54)
    print(f"{'ID':<6}{'Product':<14}{'Price (Rs.)':>12}{'Stock':>10}")
    print("=" * 54)
    for pid, info in products.items():
        print(f"{pid:<6}{info['name']:<14}{info['price']:>12,}{info['stock']:>10}")
    print("=" * 54)


def add_to_cart():
    view_products()
    pid = input("Enter Product ID to add: ").strip().upper()

    if pid not in products:                      
        print("Invalid Product ID.")
        return

    qty_text = input("Enter quantity: ").strip()
    if not qty_text.isdigit():                   
        print("Quantity must be a positive whole number.")
        return

    qty = int(qty_text)
    if qty <= 0:
        print("Quantity must be greater than 0.")
        return

    already_in_cart = cart.get(pid, 0)
    available = products[pid]["stock"]


    if already_in_cart + qty > available:
        print(f"Cannot add. Available stock = {available}, "
              f"already in cart = {already_in_cart}.")
        return

    cart[pid] = already_in_cart + qty           
    print(f"{qty} x {products[pid]['name']} added to cart. "
          f"Total in cart: {cart[pid]}")


def view_cart():
    if not cart:
        print("\nYour cart is empty.")
        return

    print("\n" + "=" * 68)
    print(f"{'ID':<6}{'Product':<14}{'Price':>12}{'Qty':>6}{'Subtotal':>14}")
    print("=" * 68)

    total = 0
    for pid, qty in cart.items():
        info = products[pid]
        subtotal = info["price"] * qty
        total += subtotal
        print(f"{pid:<6}{info['name']:<14}{info['price']:>12,}"
              f"{qty:>6}{subtotal:>14,}")

    print("=" * 68)
    print(f"{'CART TOTAL':<38}Rs. {total:>20,}")
    print("=" * 68)


def checkout():
    if not cart:
        print("\nYour cart is empty. Add items before checkout.")
        return

    total = 0
    print("\n" + "=" * 50)
    print("                 FINAL BILL")
    print("=" * 50)

    for pid, qty in cart.items():
        info = products[pid]
        subtotal = info["price"] * qty
        total += subtotal
        print(f"{info['name']:<14} x {qty:<3} = Rs. {subtotal:,}")

    print("-" * 50)
    print(f"{'Subtotal':<25}Rs. {total:,}")

    discount = 0
    if total >= 5000:                              
        discount = total * 0.10
        print(f"{'Discount (10%)':<25}Rs. {discount:,.2f}")

    final_bill = total - discount
    print(f"{'Amount to Pay':<25}Rs. {final_bill:,.2f}")
    print("=" * 50)

    for pid, qty in cart.items():
        products[pid]["stock"] -= qty
    cart.clear()

    print("Checkout successful! Thank you for shopping.")
    print("Your cart is now empty. You can start a new order.")



def main():
    while True:
        show_menu()
        choice = input("Choose an option (1-5): ").strip()

        if choice == "1":
            view_products()
        elif choice == "2":
            add_to_cart()
        elif choice == "3":
            view_cart()
        elif choice == "4":
            checkout()
        elif choice == "5":
            print("Exiting program. Goodbye!")
            break
        else:
            print("Invalid choice. Please enter a number from 1 to 5.")


main()
```
